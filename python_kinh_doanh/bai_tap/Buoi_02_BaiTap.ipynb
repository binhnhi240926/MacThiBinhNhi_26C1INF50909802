## Phần A. Bài tập có hướng dẫn

### A1. Phân hạng khách hàng bằng if/elif/else
**Quy tắc:** chi tiêu năm ≥ 50 triệu → *Kim cương*; ≥ 20 triệu → *Vàng*; ≥ 5 triệu → *Bạc*; còn lại → *Thường*.
Python xét điều kiện từ trên xuống, gặp điều kiện đúng đầu tiên thì dừng.

```python
chi_tieu = 23_500_000
if chi_tieu >= 50_000_000:
    hang = "Kim cương"
elif chi_tieu >= 20_000_000:
    hang = "Vàng"
elif chi_tieu >= 5_000_000:
    hang = "Bạc"
else:
    hang = "Thường"
print("Hạng khách hàng:", hang)
```

Kết hợp điều kiện bằng `and`, `or`, `not`:

```python
gia_tri_don, la_thanh_vien = 450_000, True
mien_phi_ship = gia_tri_don >= 500_000 or (la_thanh_vien and gia_tri_don >= 300_000)
print("Miễn phí vận chuyển?", mien_phi_ship)
```

### A2. List và vòng lặp for: tổng hợp doanh số tuần
`list` lưu một dãy giá trị có thứ tự. `for` duyệt lần lượt từng phần tử.

```python
doanh_so = [12.5, 15.2, 9.8, 18.4, 21.0, 30.5, 27.3]   # triệu đồng, Thứ 2 → Chủ nhật
ngay = ["T2", "T3", "T4", "T5", "T6", "T7", "CN"]

tong = 0
for ds in doanh_so:
    tong += ds
print(f"Tổng: {tong:.1f} – Trung bình: {tong / len(doanh_so):.2f} triệu/ngày")
print("Cao nhất:", max(doanh_so), "– Thấp nhất:", min(doanh_so), "– sum():", sum(doanh_so))

for n, ds in zip(ngay, doanh_so):          # zip ghép 2 list
    nhan = "⬆ vượt 20tr" if ds > 20 else ""
    print(f"{n}: {ds:5.1f} {nhan}")
```

`enumerate()` cho cả chỉ số; `range()` tạo dãy số; list comprehension viết vòng lặp gọn trong một dòng:

```python
for i, ds in enumerate(doanh_so, start=1):
    if i > 3: break
    print(i, ds)

ngay_vuot = [n for n, ds in zip(ngay, doanh_so) if ds > 20]
doanh_so_usd = [round(ds * 1e6 / 25_350) for ds in doanh_so]
print(ngay_vuot, doanh_so_usd)
```

### A3. Dictionary: danh mục sản phẩm
`dict` lưu cặp **khóa → giá trị**, tra cứu nhanh theo khóa. Một list các dict là cách biểu diễn bảng dữ liệu đơn giản.

```python
gia = {"SP01": 250_000, "SP02": 189_000, "SP03": 1_200_000}
gia["SP04"] = 99_000                 # thêm
gia["SP02"] = 175_000                # sửa
print(gia["SP03"], gia.get("SP99", "không có"))
for ma, g in gia.items():
    print(ma, f"{g:,}")

don_hang = [
    {"ma": "DH1", "kh": "An",   "sp": "SP01", "sl": 2},
    {"ma": "DH2", "kh": "Bình", "sp": "SP03", "sl": 1},
    {"ma": "DH3", "kh": "An",   "sp": "SP04", "sl": 5},
]
for d in don_hang:
    d["thanh_tien"] = gia[d["sp"]] * d["sl"]
print(don_hang)
```

Đếm/tổng theo nhóm bằng dict – ý tưởng đứng sau `groupby` của Pandas ở buổi 5:

```python
tong_theo_kh = {}
for d in don_hang:
    tong_theo_kh[d["kh"]] = tong_theo_kh.get(d["kh"], 0) + d["thanh_tien"]
print(tong_theo_kh)
print("Khách hàng khác nhau (set):", {d["kh"] for d in don_hang})
```

### A4. Hàm: đóng gói công thức KPI
Hàm giúp viết một lần, dùng nhiều lần. Có thể có **tham số mặc định** và **trả về nhiều giá trị** (tuple).

```python
def kpi(doanh_thu, gia_von, chi_phi_hd=0):
    """Trả về (lợi nhuận gộp, lợi nhuận ròng, biên lợi nhuận ròng)."""
    ln_gop = doanh_thu - gia_von
    ln_rong = ln_gop - chi_phi_hd
    bien = ln_rong / doanh_thu if doanh_thu else 0
    return ln_gop, ln_rong, bien

gop, rong, bien = kpi(1200, 780, 250)
print(gop, rong, f"{bien:.1%}")
print(kpi(doanh_thu=500, gia_von=300))   # dùng giá trị mặc định
```

Hàm ngắn có thể viết bằng `lambda`, thường dùng để làm khóa sắp xếp:

```python
sp = [("Áo", 250_000, 120), ("Quần", 420_000, 45), ("Mũ", 90_000, 300)]
print(sorted(sp, key=lambda x: x[1] * x[2], reverse=True))   # xếp theo doanh thu
```

### A5. while và try/except: nhập liệu an toàn
`while` lặp đến khi điều kiện sai. `try/except` bắt lỗi để chương trình không dừng đột ngột.

```python
du_lieu_nhap = ["abc", "-5", "12"]       # giả lập người dùng nhập 3 lần
i = 0
while True:
    gia_tri = du_lieu_nhap[i]; i += 1
    try:
        sl = int(gia_tri)
        if sl <= 0:
            raise ValueError("số lượng phải dương")
        break
    except ValueError as e:
        print(f"'{gia_tri}' không hợp lệ ({e}). Nhập lại.")
print("Số lượng hợp lệ:", sl)
```

## Phần B. Bài tập tự làm

### Bài B1. Hoa hồng nhân viên bán hàng

Hoa hồng tính theo doanh số tháng: dưới 100 triệu → 2%; 100–300 triệu → 3,5%; trên 300 triệu → 5% và thưởng thêm 2 triệu.
Viết hàm `hoa_hong(doanh_so)` trả về số tiền hoa hồng, rồi tính cho danh sách
`nhan_vien = {"Lan": 85e6, "Minh": 210e6, "Tuấn": 345e6, "Hà": 300e6}`.

```python
# Viết code của bạn ở đây
def hoa_hong(doanh_so):
    if doanh_so <= 100e6:
        return doanh_so*0.02
    if doanh_so <= 300e6:
        return doanh_so*0.035
    else:
        return doanh_so*0.05 + 2e6
nhan_vien = {"Lan": 85e6, "Minh": 210e6, "Tuấn": 345e6, "Hà": 300e6}
for ten,doanh_so in nhan_vien.items():
    tien_hoa_hong = hoa_hong(doanh_so)
    print(ten,":", format(tien_hoa_hong,",.0f",), "đồng")
```

### Bài B2. Phân tích danh sách đơn hàng

Cho `gia_tri_don = [320, 1250, 85, 640, 2100, 450, 99, 780, 1500, 230]` (nghìn đồng). Hãy tính:
số đơn, tổng, trung bình, số đơn ≥ 500 và tỷ lệ của chúng (%), và một list mới chỉ gồm các đơn < 100 (đơn nhỏ).

```python
# Viết code của bạn ở đây
gia_tri_don = [320, 1250, 85, 640, 2100, 450, 99, 780, 1500, 230]
so_don = len(gia_tri_don)
tong = sum(gia_tri_don)
trung_binh = tong/so_don
so_don_lon = 0
for i in gia_tri_don:
    if i >=500:
        so_don_lon +=1
ty_le_so_don_lon = so_don_lon/so_don *100
so_don_nho=[]
for gia_tri in gia_tri_don:
    if gia_tri < 100:
        so_don_nho.append(gia_tri)

print(f"Số đơn: ", so_don)
print(f"Tổng: ",tong, "nghìn đồng")
print("Trung bình: ", trung_binh, "nghìn đồng")
print(f"Số đơn lơn hơn hoặc bằng 500: ", so_don_lon, "Tỷ lệ của chúng", ty_le_so_don_lon, "%")
print("Số đơn nhỏ hơn 100: ", so_don_nho)

```

### Bài B3. Giỏ hàng

Cho bảng giá `gia = {"Cà phê": 29000, "Trà sữa": 35000, "Bánh mì": 20000, "Nước suối": 10000}` và giỏ hàng
`gio = [("Cà phê", 2), ("Bánh mì", 3), ("Trà đào", 1), ("Nước suối", 4)]`.
Tính tổng tiền; món không có trong bảng giá thì in cảnh báo và bỏ qua. Nếu tổng ≥ 150.000 đ giảm 10%.

```python
# Viết code của bạn ở đây
gia = {"Cà phê": 29000, "Trà sữa": 35000, "Bánh mì": 20000, "Nước suối": 10000}
gio = [("Cà phê", 2), ("Bánh mì", 3), ("Trà đào", 1), ("Nước suối", 4)]
tong_tien = 0
for sp,sl in gio :
    if sp in gia:
        thanh_tien = sl*gia[sp]
        tong_tien += thanh_tien
    else:
        print(f"Cảnh báo",sp, "không có trong danh sách nên hãy bỏ qua.")
print(f"Tổng tiền trước giảm: {tong_tien:.0f} vnd")
if tong_tien >= 150_000:
    gia_giam = tong_tien * 0.1
else:
    gia_giam = 0
tong_tien_sau_giam = tong_tien - gia_giam
print(f"Tổng tiền sau giảm {tong_tien_sau_giam:.0f} vnd")
```

### Bài B4. Thống kê theo nhóm

Cho danh sách giao dịch `gd = [("Điện tử", 12.5), ("Thời trang", 3.2), ("Điện tử", 8.1), ("Thực phẩm", 1.5),
("Thời trang", 4.8), ("Điện tử", 15.0), ("Thực phẩm", 2.2)]` (triệu đồng).
Dùng dict để tính tổng doanh thu và số giao dịch mỗi nhóm hàng; in nhóm có doanh thu cao nhất.

```python
# Viết code của bạn ở đây
gd = [("Điện tử", 12.5), ("Thời trang", 3.2), ("Điện tử", 8.1), ("Thực phẩm", 1.5), ("Thời trang", 4.8), ("Điện tử", 15.0), ("Thực phẩm", 2.2)]
doanh_thu = {}
so_giao_dich = {}
for nhom, tien in gd:
    if nhom not in doanh_thu:
        doanh_thu[nhom] = 0
        so_giao_dich[nhom] = 0
    doanh_thu[nhom] += tien
    so_giao_dich[nhom] += 1
print("Thống kê theo nhóm hàng:")
for nhom in doanh_thu:
    print(nhom,"- Doanh thu:", doanh_thu[nhom], "triệu đồng","- Số giao dịch:", so_giao_dich[nhom])
nhom_cao_nhat = max(doanh_thu, key=doanh_thu.get)

print("Nhóm có doanh thu cao nhất:", nhom_cao_nhat)
print("Doanh thu:", doanh_thu[nhom_cao_nhat], "triệu đồng")



```

### Bài B5. Kế hoạch tiết kiệm với while

Một startup có 500 triệu đồng tiền mặt, mỗi tháng tiêu (burn) 65 triệu đồng và chi phí tăng 3% mỗi tháng.
Dùng `while` để tìm số tháng công ty còn hoạt động được trước khi hết tiền (runway).

```python
# Viết code của bạn ở đây
tien_mat = 500
chi_phi = 65
so_thang = 0

while tien_mat >= chi_phi:
    tien_mat -= chi_phi
    so_thang += 1
    chi_phi *= 1.03
print("Runway:", so_thang, "tháng")
print("Tiền còn lại:", round(tien_mat, 2), "triệu đồng")
```

## Phần C. Bài tập làm thêm

### Bài C1. Hàm chuyển số tiền thành chữ (rút gọn)

Viết hàm `rut_gon(x)` hiển thị số tiền dạng rút gọn: `1_250_000_000 → "1,25 tỷ"`, `45_600_000 → "45,6 triệu"`,
`850_000 → "850 nghìn"`, `900 → "900 đ"`. Thử với một list nhiều số.

```python
# Viết code của bạn ở đây
def rut_gon(x):
    if x >= 1_000_000_000:
        return str(round(x / 1_000_000_000, 2)).replace(".", ",") + " tỷ"
    elif x >= 1_000_000:
        return str(round(x / 1_000_000, 2)).replace(".", ",") + " triệu"
    elif x >= 1_000:
        return str(round(x / 1_000, 2)).replace(".", ",") + " nghìn"
    else:
        return str(x) + " đ"
danh_sach = [
    1_250_000_000,
    45_600_000,
    850_000,
    900,
    2_500_000_000,
    75_000,
    1_200
]
for so_tien in danh_sach:
    print(rut_gon(so_tien))
```

### Bài C2. Tồn kho tối thiểu

Cho `kho = {"SP01": {"ton": 12, "ban_tb_ngay": 3, "giao_hang_ngay": 5}, "SP02": {"ton": 80, "ban_tb_ngay": 6, "giao_hang_ngay": 7},
"SP03": {"ton": 20, "ban_tb_ngay": 2, "giao_hang_ngay": 4}}`.
Điểm đặt hàng lại = bán TB/ngày × thời gian giao hàng × 1,2 (hệ số an toàn). Liệt kê các sản phẩm cần đặt hàng ngay
(tồn ≤ điểm đặt hàng lại), sắp xếp theo số ngày còn bán được tăng dần.

```python
# Viết code của bạn ở đây
kho = {
    "SP01": {"ton": 12, "ban_tb_ngay": 3, "giao_hang_ngay": 5},
    "SP02": {"ton": 80, "ban_tb_ngay": 6, "giao_hang_ngay": 7},
    "SP03": {"ton": 20, "ban_tb_ngay": 2, "giao_hang_ngay": 4}
}

can_dat_hang = []

for ma_sp, thong_tin in kho.items():
    ton = thong_tin["ton"]
    ban_tb_ngay = thong_tin["ban_tb_ngay"]
    giao_hang_ngay = thong_tin["giao_hang_ngay"]

    diem_dat_hang = ban_tb_ngay * giao_hang_ngay * 1.2
    so_ngay_con_ban = ton / ban_tb_ngay

    if ton <= diem_dat_hang:
        can_dat_hang.append(
            (ma_sp, ton, diem_dat_hang, so_ngay_con_ban)
        )

can_dat_hang.sort(key=lambda x: x[3])

print("Các sản phẩm cần đặt hàng ngay:")

for ma_sp, ton, diem_dat_hang, so_ngay in can_dat_hang:
    print("Mã sản phẩm:", ma_sp)
    print("Tồn kho:", ton)
    print("Điểm đặt hàng lại:", round(diem_dat_hang, 2))
    print("Số ngày còn bán được:", round(so_ngay, 2))
    print()

```

### Bài C3. Mô phỏng doanh thu với số ngẫu nhiên

Dùng `random` mô phỏng 30 ngày bán hàng: mỗi ngày có từ 50 đến 120 khách (`randint`), mỗi khách chi trung bình
ngẫu nhiên từ 80.000 đến 250.000 đ (`uniform`). Tính tổng doanh thu tháng, ngày cao nhất, và lặp mô phỏng 1.000 lần
để ước lượng xác suất doanh thu tháng vượt 450 triệu đồng. Dùng `random.seed(1)` để kết quả lặp lại được.

```python
# Viết code của bạn ở đây

import random

random.seed(1)

# Phần 1: Mô phỏng 30 ngày bán hàng
doanh_thu_ngay = []

for ngay in range(30):
    so_khach = random.randint(50, 120)
    chi_tieu_tb = random.uniform(80000, 250000)

    doanh_thu = so_khach * chi_tieu_tb
    doanh_thu_ngay.append(doanh_thu)

tong_doanh_thu = sum(doanh_thu_ngay)
doanh_thu_cao_nhat = max(doanh_thu_ngay)
ngay_cao_nhat = doanh_thu_ngay.index(doanh_thu_cao_nhat) + 1

print("Tổng doanh thu tháng:", format(round(tong_doanh_thu), ",.0f"), "đ")
print("Ngày doanh thu cao nhất:", ngay_cao_nhat)
print("Doanh thu ngày cao nhất:", format(round(doanh_thu_cao_nhat), ",.0f"), "đ")

# Phần 2: Mô phỏng 1.000 tháng
so_thang_vuot_muc = 0

for lan in range(1000):
    tong_thang = 0

    for ngay in range(30):
        so_khach = random.randint(50, 120)
        chi_tieu_tb = random.uniform(80000, 250000)

        tong_thang += so_khach * chi_tieu_tb

    if tong_thang > 450_000_000:
        so_thang_vuot_muc += 1

xac_suat = so_thang_vuot_muc / 1000 * 100

print("Số tháng vượt 450 triệu:", so_thang_vuot_muc)
print("Xác suất ước lượng:", round(xac_suat, 2), "%")
```
