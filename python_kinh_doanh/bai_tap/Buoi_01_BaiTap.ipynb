### A2. Tính giá bán sau chiết khấu và thuế VAT
**Tình huống:** Cửa hàng bán một chiếc tai nghe giá niêm yết 850.000 đ, giảm 15%, sau đó cộng VAT 8%. Khách mua 3 chiếc.

**Bước 1 – Khai báo biến.** Tên biến viết thường, nối bằng dấu `_`, đặt tên có nghĩa.

```python
gia_niem_yet = 850_000      # đồng (dấu _ giúp dễ đọc số lớn)
ty_le_giam = 0.15           # 15%
vat = 0.08                  # 8%
so_luong = 3
print(type(gia_niem_yet), type(ty_le_giam), type(so_luong))
```

**Bước 2 – Tính toán** với các toán tử `+ - * /`.

```python
gia_sau_giam = gia_niem_yet * (1 - ty_le_giam)
gia_sau_vat = gia_sau_giam * (1 + vat)
thanh_tien = gia_sau_vat * so_luong
print(gia_sau_giam, gia_sau_vat, thanh_tien)
```

**Bước 3 – Định dạng kết quả.** f-string `f"...{bien:,.0f}..."` chèn giá trị vào chuỗi;
`:,.0f` = dấu phẩy ngăn cách hàng nghìn, 0 chữ số thập phân. Thay `,` bằng `.` để đúng kiểu Việt Nam.

```python
def vnd(x):
    return f"{x:,.0f}".replace(",", ".") + " đ"

print("Giá sau giảm :", vnd(gia_sau_giam))
print("Giá sau VAT  :", vnd(gia_sau_vat))
print("Thành tiền   :", vnd(thanh_tien))
```

### A3. Phép chia lấy nguyên `//`, chia lấy dư `%` và lũy thừa `**`
**Tình huống:** Kho có 1.250 chai nước, mỗi thùng đóng 24 chai. Được bao nhiêu thùng, dư bao nhiêu chai?

```python
tong_chai = 1250
chai_moi_thung = 24
so_thung = tong_chai // chai_moi_thung
du = tong_chai % chai_moi_thung
print(f"Đóng được {so_thung} thùng, còn dư {du} chai")
```

**Lãi kép:** Gửi 100 triệu đồng, lãi suất 6%/năm, sau 5 năm được bao nhiêu? Công thức: `A = P × (1 + r)^n`.

```python
P, r, n = 100_000_000, 0.06, 5
A = P * (1 + r) ** n
print("Số tiền sau", n, "năm:", vnd(A))
print("Tiền lãi:", vnd(A - P))
```

### Bài B1. Hóa đơn bán hàng

Khách mua 2 áo sơ mi (350.000 đ/chiếc) và 1 quần tây (520.000 đ). Cửa hàng giảm 10% trên tổng tiền hàng,
sau đó tính VAT 8%. In ra: tổng tiền hàng, tiền giảm, tiền VAT, số tiền khách phải trả (định dạng `1.234.567 đ`).

```python
# Viết code của bạn ở đây
# Giá và số lượng sản phẩm
gia_ao_so_mi = 350_000
so_luong_so_mi = 2
gia_quan_tay = 520_000
so_luong_quan_tay = 1

# Tỷ lệ giảm giá và VAT
ty_le_giam = 0.10
vat = 0.08

# 1. Tổng tiền hàng
tong_tien_hang = gia_ao_so_mi * so_luong_so_mi + gia_quan_tay * so_luong_quan_tay

# 2. Tiền giảm giá
tien_giam = tong_tien_hang * ty_le_giam

# 3. Tiền hàng sau giảm giá
tien_sau_giam = tong_tien_hang - tien_giam

# 4. Tiền VAT (tính sau khi giảm giá)
tien_vat = tien_sau_giam * vat

# 5. Số tiền khách phải trả
thanh_tien = tien_sau_giam + tien_vat

# 6. In kết quả theo định dạng 1.234.567 đ
print(f"Tổng tiền hàng: {tong_tien_hang:,.0f} đ".replace(",", "."))
print(f"Tiền giảm: {tien_giam:,.0f} đ".replace(",", "."))
print(f"Tiền VAT: {tien_vat:,.0f} đ".replace(",", "."))
print(f"Số tiền khách phải trả: {thanh_tien:,.0f} đ".replace(",", "."))

```

### Bài B2. Điểm hòa vốn

Một quán cà phê có định phí 45 triệu đồng/tháng, giá bán trung bình 35.000 đ/ly, biến phí 14.000 đ/ly.
Tính số ly cần bán mỗi tháng để hòa vốn (làm tròn lên) và số ly mỗi ngày (tháng 30 ngày).
Công thức: `Q = Định phí / (Giá bán − Biến phí)`.

<details><summary>💡 Gợi ý</summary>

Dùng `math.ceil()` để làm tròn lên (cần `import math`).

</details>

```python

import math

# Khai báo dữ liệu
dinh_phi = 45_000_000
gia_ban = 35_000
bien_phi = 14_000
so_ngay = 30

# 1. Số ly cần bán mỗi tháng để hòa vốn
so_ly_thang = math.ceil(dinh_phi / (gia_ban - bien_phi))

# 2. Số ly cần bán trung bình mỗi ngày
so_ly_ngay = math.ceil(so_ly_thang / so_ngay)

# In kết quả
print(f"Số ly cần bán mỗi tháng: {so_ly_thang:,} ly".replace(",", "."))
print(f"Số ly cần bán mỗi ngày: {so_ly_ngay:,} ly".replace(",", "."))
# Viết code của bạn ở đây
```

### Bài B3. Trả góp hàng tháng

Khách mua laptop 24.000.000 đ trả góp 12 tháng, lãi suất 1,2%/tháng. Tính số tiền trả mỗi tháng theo công thức
niên kim: `PMT = P × r × (1 + r)^n / ((1 + r)^n − 1)`, tổng số tiền trả và tổng tiền lãi.

```python
# Viết code của bạn ở đây

# Khai báo dữ liệu
P = 24_000_000       # Số tiền vay (đồng)
r = 0.012            # Lãi suất 1,2%/tháng
n = 12               # Thời hạn trả góp (tháng)

# Tính số tiền trả mỗi tháng theo công thức niên kim
pmt = P * r * (1 + r)**n / ((1 + r)**n - 1)

# Tính tổng số tiền trả và tổng tiền lãi
tong_tien_tra = pmt * n
tong_tien_lai = tong_tien_tra - P

# In kết quả
print(f"Số tiền trả mỗi tháng: {pmt:,.0f} đ".replace(",", "."))
print(f"Tổng số tiền trả: {tong_tien_tra:,.0f} đ".replace(",", "."))
print(f"Tổng tiền lãi: {tong_tien_lai:,.0f} đ".replace(",", "."))
```

### Bài B5. Tăng trưởng doanh thu

Doanh thu quý 1 là 1,85 tỷ đồng, quý 2 là 2,12 tỷ đồng. Tính tỷ lệ tăng trưởng (%) và in dạng `Tăng trưởng: 14.6%`.
Nếu duy trì tốc độ này, doanh thu quý 4 dự kiến bao nhiêu?

```python
# Viết code của bạn ở đây
doanh_thu_q1 = 1.85
doanh_thu_q2 = 2.12
tang_truong = (doanh_thu_q2 - doanh_thu_q1) / doanh_thu_q1 * 100
doanh_thu_q3 = doanh_thu_q2 * (1 + tang_truong / 100)
doanh_thu_q4 = doanh_thu_q3 * (1 + tang_truong / 100)
# In kết quả
print(f"Tăng trưởng: {tang_truong:.1f}%")
print(f"Doanh thu quý 4 dự kiến: {doanh_thu_q4:.2f} tỷ đồng")
```

### Bài C1. So sánh hai phương án gửi tiết kiệm

Phương án 1: lãi 6,2%/năm, nhập lãi hằng năm. Phương án 2: lãi 6,0%/năm nhưng nhập lãi hằng tháng
(lãi tháng = 6,0%/12). Gửi 200 triệu đồng trong 3 năm. Phương án nào có lợi hơn và hơn bao nhiêu?

```python
# Viết code của bạn ở đây
tien_gui = 200_000_000
so_nam = 3
# Phương án 1: Lãi 6,2%/năm, nhập lãi hằng năm
lai_suat_1 = 0.062
tien_cuoi_1 = tien_gui * (1 + lai_suat_1) ** so_nam

# Phương án 2: Lãi 6,0%/năm, nhập lãi hằng tháng
lai_suat_2 = 0.06
lai_suat_thang = lai_suat_2 / 12
so_thang = so_nam * 12
tien_cuoi_2 = tien_gui * (1 + lai_suat_thang) ** so_thang

# So sánh hai phương án
chenh_lech = abs(tien_cuoi_1 - tien_cuoi_2)

# In kết quả
print(f"Phương án 1: {tien_cuoi_1:,.0f} đ".replace(",", "."))
print(f"Phương án 2: {tien_cuoi_2:,.0f} đ".replace(",", "."))
print(f"Chênh lệch: {chenh_lech:,.0f} đ".replace(",", "."))

if tien_cuoi_1 > tien_cuoi_2:
    print("Phương án 1 có lợi hơn.")
elif tien_cuoi_2 > tien_cuoi_1:
    print("Phương án 2 có lợi hơn.")
else:
    print("Hai phương án có lợi ích như nhau.")
```

### Bài C2. Đọc thông tin từ chuỗi đơn hàng

Cho chuỗi `don = "DH2025-1123|Nguyễn Thị Hoa|0903123456|3|189000"` (mã đơn | tên | SĐT | số lượng | đơn giá).
Tách các trường, che số điện thoại thành `0903***456` và in hóa đơn gọn gàng gồm thành tiền.

```python
# Viết code của bạn ở đây

# Chuỗi thông tin đơn hàng
don = "DH2025-1123|Nguyễn Thị Hoa|0903123456|3|189000"

# 1. Tách các trường thông tin
ma_don, ten_khach, sdt, so_luong, don_gia = don.split("|")

# 2. Chuyển kiểu dữ liệu
so_luong = int(so_luong)
don_gia = int(don_gia)

# 3. Che số điện thoại: giữ 4 số đầu và 3 số cuối
sdt_che = sdt[:4] + "***" + sdt[-3:]

# 4. Tính thành tiền
thanh_tien = so_luong * don_gia

# 5. In hóa đơn
print("==== HÓA ĐƠN BÁN HÀNG ====")
print(f"Mã đơn:       {ma_don}")
print(f"Khách hàng:   {ten_khach}")
print(f"Số điện thoại: {sdt_che}")
print(f"Số lượng:     {so_luong}")
print(f"Đơn giá:      {don_gia:,} đ".replace(",", "."))
print("----------------------------")
print(f"Thành tiền:   {thanh_tien:,} đ".replace(",", "."))
print("============================")
```

### Bài C3. Quy tắc 72

Quy tắc 72 ước lượng số năm để vốn tăng gấp đôi: `72 / lãi suất (%)`. Với lãi suất 4%, 6%, 8%, 12%,
hãy so sánh kết quả ước lượng với kết quả chính xác `ln(2) / ln(1 + r)`. (Có thể viết lặp lại 4 lần;
buổi sau ta sẽ dùng vòng lặp.)

```python
# Viết code của bạn ở đây

import math

# Lãi suất theo phần trăm
lai_suat_1 = 4
lai_suat_2 = 6
lai_suat_3 = 8
lai_suat_4 = 12

# Lãi suất đổi sang số thập phân
r1 = lai_suat_1 / 100
r2 = lai_suat_2 / 100
r3 = lai_suat_3 / 100
r4 = lai_suat_4 / 100

# Quy tắc 72
uoc_luong_1 = 72 / lai_suat_1
uoc_luong_2 = 72 / lai_suat_2
uoc_luong_3 = 72 / lai_suat_3
uoc_luong_4 = 72 / lai_suat_4

# Công thức chính xác
chinh_xac_1 = math.log(2) / math.log(1 + r1)
chinh_xac_2 = math.log(2) / math.log(1 + r2)
chinh_xac_3 = math.log(2) / math.log(1 + r3)
chinh_xac_4 = math.log(2) / math.log(1 + r4)

# In kết quả
print(f"Lãi suất {lai_suat_1}%: Quy tắc 72 = {uoc_luong_1:.2f} năm, Chính xác = {chinh_xac_1:.2f} năm")
print(f"Lãi suất {lai_suat_2}%: Quy tắc 72 = {uoc_luong_2:.2f} năm, Chính xác = {chinh_xac_2:.2f} năm")
print(f"Lãi suất {lai_suat_3}%: Quy tắc 72 = {uoc_luong_3:.2f} năm, Chính xác = {chinh_xac_3:.2f} năm")
print(f"Lãi suất {lai_suat_4}%: Quy tắc 72 = {uoc_luong_4:.2f} năm, Chính xác = {chinh_xac_4:.2f} năm")
```
